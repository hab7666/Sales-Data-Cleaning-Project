**Import Libraries and Load DatasetPython**

```python
import pandas as pd
import numpy as np
import re

df = pd.read_csv('Messy_data.csv')

print("Dataset loaded successfully")
print("Shape:", df.shape)
df.head(10)
```

Initial Inspection

```python
print("=== Dataset Info ===")
print(df.info())

print("\n=== Missing Values ===")
print(df.isnull().sum())

print("\n=== Duplicate Rows ===")
print("Total exact duplicates:", df.duplicated().sum())
```

Remove Extra Spaces

```python
df['CustomerName'] = df['CustomerName'].str.strip()
df['City'] = df['City'].str.strip()
df['Product'] = df['Product'].str.strip()
df['OrderStatus'] = df['OrderStatus'].str.strip()

print("Extra spaces removed")
df[['CustomerName', 'City', 'Product', 'OrderStatus']].head(8)
```

Standardize Text

```python
df['City'] = df['City'].str.title()
df['Product'] = df['Product'].str.title()
df['OrderStatus'] = df['OrderStatus'].str.title()

print("Text standardized")
print("\nUnique Cities:", df['City'].unique())
print("\nUnique Order Status:", df['OrderStatus'].unique())
print("\nUnique Products:", df['Product'].unique())
```

Clean Quantity Column

```python
def clean_quantity(val):
    if pd.isna(val):
        return np.nan
    val = str(val).lower().strip()

    word_map = {'three': 3, 'two': 2, 'four': 4, 'five': 5}
    if val in word_map:
        return word_map[val]

    match = re.search(r'(-?\d+)', val)
    if match:
        return int(match.group(1))

    return np.nan

df['Quantity'] = df['Quantity'].apply(clean_quantity)

print("Quantity column cleaned")
print(df['Quantity'].unique())
```

Clean UnitPrice Column

```python
df['UnitPrice'] = df['UnitPrice'].astype(str).str.replace(',', '', regex=False)
df['UnitPrice'] = df['UnitPrice'].replace('unknown', np.nan)
df['UnitPrice'] = pd.to_numeric(df['UnitPrice'], errors='coerce')

print("UnitPrice column cleaned")
print(df['UnitPrice'].describe())
```

Clean CustomerAge Column

```python
def clean_age(val):
    if pd.isna(val):
        return np.nan
    val = str(val).lower().strip()

    if val == 'forty':
        return 40
    if 'years' in val:
        match = re.search(r'(\d+)', val)
        if match:
            return int(match.group(1))

    try:
        return float(val)
    except:
        return np.nan

df['CustomerAge'] = df['CustomerAge'].apply(clean_age)

print("CustomerAge column cleaned")
print(df['CustomerAge'].unique()[:15])
```

Clean OrderDate Column

```python
df['OrderDate'] = pd.to_datetime(df['OrderDate'], errors='coerce', dayfirst=True)

print("OrderDate column cleaned")
print("Invalid dates (NaT):", df['OrderDate'].isna().sum())
df[['OrderID', 'OrderDate']].head(10)
```

Handle Missing Values

```python
print("Missing values before handling:")
print(df.isnull().sum())

df['CustomerName'] = df['CustomerName'].fillna('Unknown')
df['City'] = df['City'].fillna(df['City'].mode()[0])
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())
df['UnitPrice'] = df['UnitPrice'].fillna(df['UnitPrice'].median())
df['CustomerAge'] = df['CustomerAge'].fillna(df['CustomerAge'].median())
df['OrderStatus'] = df['OrderStatus'].fillna(df['OrderStatus'].mode()[0])

# Drop rows with invalid dates
df = df.dropna(subset=['OrderDate'])

print("\nMissing values after handling:")
print(df.isnull().sum())
```

Handle Outliers

```python
# Keep realistic age range
df = df[(df['CustomerAge'] >= 10) & (df['CustomerAge'] <= 100)]

# Replace extreme Quantity values
quantity_median = df['Quantity'].median()
df.loc[df['Quantity'] > 50, 'Quantity'] = quantity_median
df.loc[df['Quantity'] < 0, 'Quantity'] = quantity_median

# Replace extreme UnitPrice values
price_median = df['UnitPrice'].median()
df.loc[df['UnitPrice'] > 200000, 'UnitPrice'] = price_median
df.loc[df['UnitPrice'] < 1, 'UnitPrice'] = price_median

print("Outliers handled")
print("Quantity range:", df['Quantity'].min(), "to", df['Quantity'].max())
print("Age range:", df['CustomerAge'].min(), "to", df['CustomerAge'].max())
print("Max UnitPrice:", df['UnitPrice'].max())
```

Remove Duplicate RecordsPython

```python
print("Duplicates before:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates after:", df.duplicated().sum())
print("Final shape:", df.shape)
```

Convert Data Types

```python
df['OrderID'] = df['OrderID'].astype(int)
df['Quantity'] = df['Quantity'].astype(int)
df['CustomerAge'] = df['CustomerAge'].astype(int)
df['UnitPrice'] = df['UnitPrice'].astype(float)

print("Data types converted")
print(df.dtypes)
```

Create TotalAmount Column

```python
df['TotalAmount'] = df['Quantity'] * df['UnitPrice']

print("TotalAmount column created")
df[['Quantity', 'UnitPrice', 'TotalAmount']].head(10)
```

Final Check and Export

```python
print("=== FINAL CLEANED DATASET ===")
print("Shape:", df.shape)
print("\nMissing values:")
print(df.isnull().sum())
print("\nFirst 10 rows:")
display(df.head(10))

df.to_csv('cleaned_sales_dataset.csv', index=False)
print("\nCleaned dataset saved as 'cleaned_sales_dataset.csv'")
```
